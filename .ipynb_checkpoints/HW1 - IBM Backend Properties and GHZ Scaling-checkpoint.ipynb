```python
# ---------------------------------
# READING BACKEND CHARACTERIZATIONS
# ---------------------------------
device_info = {
    "ibm_fez": {"T1": [], "T2": [], "Readout": []},
    "ibm_marrakesh": {"T1": [], "T2": [], "Readout": []},
}

num_qubits_b1 = backend_one.num_qubits
num_qubits_b2 = backend_two.num_qubits
print("IBM_FEZ: ",num_qubits_b1)
print("IBM_MARRAKESH: ",num_qubits_b2)

backend_one_prop = backend_one.properties(refresh=True)
backend_two_prop = backend_two.properties(refresh=True)

# Reading error from IBM_FEZ
for qubit in range(num_qubits_b1):
    try:
        device_info["ibm_fez"]["T1"].append(backend_one_prop.t1(qubit) * 1e6)
    except Exception:
        device_info["ibm_fez"]["T1"].append(np.nan)
    try:
        device_info["ibm_fez"]["T2"].append(backend_one_prop.t2(qubit) * 1e6)
    except Exception:
        device_info["ibm_fez"]["T2"].append(np.nan)
    try:
        device_info["ibm_fez"]["Readout"].append(backend_one_prop.readout_error(qubit))
    except Exception:
        device_info["ibm_fez"]["Readout"].append(np.nan)

# # Reading error from IBM_MERRAKESH
# for qubit in range(num_qubits_b2):
#     device_info["ibm_marrakesh"]["T1"].append(backend_two_prop.t1(qubit) * 1e6)
#     device_info["ibm_marrakesh"]["T2"].append(backend_two_prop.t2(qubit) * 1e6)
#     device_info["ibm_marrakesh"]["Readout"].append(backend_two_prop.readout_error(qubit))
print("hi")
    
   
fez_df = pd.DataFrame(device_info["ibm_fez"])
fez_df
```
